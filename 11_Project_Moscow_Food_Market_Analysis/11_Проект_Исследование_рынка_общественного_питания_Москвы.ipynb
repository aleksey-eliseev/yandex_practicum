Файл `moscow_places.csv`:

- `name` — название заведения;
- `address` — адрес заведения;
- `category` — категория заведения, например «кафе», «пиццерия» или «кофейня»;
- `hours` — информация о днях и часах работы;
- `lat` — широта географической точки, в которой находится заведение;
- `lng` — долгота географической точки, в которой находится заведение;
- `rating` — рейтинг заведения по оценкам пользователей в Яндекс Картах (высшая оценка — `5.0`);
- `price` — категория цен в заведении, например «средние», «ниже среднего», «выше среднего» и так далее;
- `avg_bill` — строка, которая хранит среднюю стоимость заказа в виде диапазона, например:
    - «Средний счёт: 1000–1500 ₽»;
    - «Цена чашки капучино: 130–220 ₽»;
    - «Цена бокала пива: 400–600 ₽».
        
        и так далее;
        
- `middle_avg_bill` — число с оценкой среднего чека, которое указано только для значений из столбца `avg_bill`, начинающихся с подстроки «Средний счёт»:
    - Если в строке указан ценовой диапазон из двух значений, в столбец войдёт медиана этих двух значений.
    - Если в строке указано одно число — цена без диапазона, то в столбец войдёт это число.
    - Если значения нет или оно не начинается с подстроки «Средний счёт», то в столбец ничего не войдёт.
- `middle_coffee_cup` — число с оценкой одной чашки капучино, которое указано только для значений из столбца `avg_bill`, начинающихся с подстроки «Цена одной чашки капучино»:
    - Если в строке указан ценовой диапазон из двух значений, в столбец войдёт медиана этих двух значений.
    - Если в строке указано одно число — цена без диапазона, то в столбец войдёт это число.
    - Если значения нет или оно не начинается с подстроки «Цена одной чашки капучино», то в столбец ничего не войдёт.
- `chain` — число, выраженное `0` или `1`, которое показывает, является ли заведение сетевым (для маленьких сетей могут встречаться ошибки);
- `district` — административный район, в котором находится заведение, например Центральный административный округ;
- `seats` — количество посадочных мест.

## Предобработка данных

Выполним предобработку данных

### Пропуски и дубликаты

```python
# Приведём все строковые данные к нижнему регистру
for col in ['name', 'category', 'address', 'hours', 'price', 'avg_bill']:
    places[col] = places[col].str.lower()
```

```python
# Проверим датафрейм на наличие полных и частичных дубликатов с дублями по 'name' и 'address'
print(f'Кол-во полных дубликатов:\n{places.duplicated().sum()}\n')
print(f'Кол-во частичных дубликатов:\n{places.duplicated(subset = ["name", "address"]).sum()}\n')
```

```python
# Выведем строки с дубликатами
places.loc[places.duplicated(subset = ["name", "address"], keep=False)]
```

Был обнаружен ряд дубликатов с одинаковыми названиями и адресами, разберёмся с каждым по отдельности.

- **"кафе"** — небольшие различия в часах работы и координатах. Вероятно заведение переехало в пределах той-же улицы. Удалим дубликат.
- **"more poke"** — различия в часах работы и принадлежности к сети. Интернет говорит, что данное заведение принадлежит к сети. Оставим вариант с указанием принадлежности к сети.
- **"раковарня клешни и хвосты"** — так же принадлежит к сети. Оставляем соответствующий вариант.
- **"хлеб да выпечка"** — к сети не принадлежит. Удалим соответствующий вариант.

```python
# Удалим соответствующие дубликаты

# "more poke" — оставить запись с chain = 1
places = places[~((places['name'] == 'more poke') & (places['chain'] == 0))]

# "раковарня клешни и хвосты" — оставить запись с chain = 1
places = places[~((places['name'] == 'раковарня клешни и хвосты') & (places['chain'] == 0))]

# "хлеб да выпечка" — оставить запись с chain = 0
places = places[~((places['name'] == 'хлеб да выпечка') & (places['chain'] == 1))]

places = places.drop_duplicates(subset=['name', 'address'])
```

```python
# Проверим датафрейм на наличие пропусков
print(f'Кол-во пропусков:\n{places.isna().sum()}\n')
```

Пропусков достаточно много. Просто удалить их не получится, иначе достоверность исследования окажется под сомнением. Попробуем исправит ьих по ходу исследования.

### Добавление столбцов

```python
# Добавим столбец `street` с названиями улиц из столбца с адресом
places['street'] = places['address'].str.extract(r',\s*([^,]+),')
```

### Количество посадочных мест по категориям заведений

Исследуем количество посадочных мест в заведениях по категориям: рестораны, кофейни, пиццерии, бары и так далее. Построим визуализацию.

Для начала посмотрим медианное количество посадочных мест, приходящихся на категорию заведений.

```python
# Посчитаем медианное количество посадочных мест на категорию заведения
category_seats_median = (
    places.groupby('category', as_index=False)
    .agg(seats_median=('seats', 'median'))
    .sort_values('seats_median', ascending=False)
    .reset_index(drop=True)
)

category_seats_median
```

Ожидаемо больше всего посадочных мест приходится на рестораны, меньше всего на булочные.

```python
# Боксплот распределения количества мест по категориям
category_order = category_seats_median['category']

fig = px.box(
    places,
    x='seats',
    y='category',
    title='Распределение количества посадочных мест по категориям',
    labels={'seats': 'Количество мест', 'category': 'Категория'},
    color='category',
    template='plotly_white',
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_layout(
    yaxis=dict(categoryorder='array', categoryarray=list(category_order)),
    xaxis_title='Количество мест',
    yaxis_title='Категория',
    title_x=0.5
)

fig.show()
```

На графике видны подозрительные выбросы с одинаково большим количеством посадочных мест в разных категориях заведений. Исследуем этот вопрос. Возможно эти заведения находятся в торговых центрах и количество мест указывается одинаковое на весь фудкорт.

```python
# Посмотрим на заведения с более чем 1000 посадочных мест.
places.query('seats > 1000')
```

Гипотеза подтвердилась частично, заведения распологающиеся по адресу **"Москва, Измайловское шоссе, 71, корп. А"** и с количеством посадочных мест **1040** действительно распологаются в торговом центре. При этом остальные заведения с количеством мест **1288** располагаются по разным адресам, хотя и на одной улице. Вероятно значения посадочных мест были проставлены ошибочно.

```python
# Избавимся от выбросов и пострпоим боксплот заново.
Q1 = places['seats'].quantile(0.25)
Q3 = places['seats'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
places_new = places[(places['seats'] >= lower_bound) & (places['seats'] <= upper_bound)]

# График распределения количества мест по категориям без выбросов
fig = px.box(
    places_new,
    x='seats',
    y='category',
    title='Распределение количества посадочных мест по категориям (без выбросов)',
    labels={'seats': 'Количество мест', 'category': 'Категория'},
    color='category',
    template='plotly_white',
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_layout(
    yaxis=dict(categoryorder='array', categoryarray=list(category_order)),
    xaxis_title='Количество мест',
    yaxis_title='Категория',
    title_x=0.5
)

fig.show()
```

---

**Вывод:**<br/>
- **Рестораны** имеют самое широкое распределение по количеству посадочных мест (**медиана 80** мест). Это подтверждает, что рестораны, как правило, имеют больше места для гостей по сравнению с другими категориями.

### Круглосуточные кофейни

Определим количество круглосуточных кофеен в датасете.

```python
# Фильтруем круглосуточные кофейни
coffee_shops_24 = coffee_shops[coffee_shops['is_24_7']]

print(f"Количество круглосуточных кофеен: {len(coffee_shops_24)}")
```

```python
# Определим в каких округах расположены круглосуточные кофейни
coffee_shops_24_district = (
    coffee_shops_24.groupby('district') # название района
    .size()
    .reset_index(name='count') # кол-во круглосуточных кофеен в районе
)

# Рассчитаем долю круглосуточных кофеен по округам
coffee_shops_total_district = coffee_shops.groupby('district').size().reset_index(name='total_count') # общее кол-во кофеен в районе
coffee_shops_24_district = coffee_shops_24_district.merge(coffee_shops_total_district, on='district')
coffee_shops_24_district['percentage'] = (coffee_shops_24_district['count'] / coffee_shops_24_district['total_count']) * 100 # процент круглосуточных кофеен от общего числа

coffee_shops_24_district = coffee_shops_24_district.sort_values(by='count', ascending=False).reset_index(drop=True)

coffee_shops_24_district
```

```python
# Строим столбчатую диаграмму для отображения доли круглосуточных кофеен по районам
coffee_shops_24_district = coffee_shops_24_district.sort_values(by='percentage', ascending=False)

fig = px.bar(
    coffee_shops_24_district,
    x='district', 
    y='percentage', 
    title='Доля круглосуточных кофеен по районам',
    labels={'district': 'Район', 'percentage': 'Доля круглосуточных кофеен (%)'},
    color='percentage',
    color_continuous_scale='Viridis',
    text='percentage'
)

# Настроим отображение текста и оформление графика
fig.update_traces(texttemplate='%{text:.2f}%', textposition='outside')
fig.update_layout(
    xaxis_title='Район',
    yaxis_title='Доля круглосуточных кофеен (%)',
    title_x=0.5,
    xaxis_tickangle=45,
    showlegend=False
)

# Покажем график
fig.show()
```

### Распределение рейтинга кофеен

Определим распределение рейтинга кофеен по районам.

```python
# Группируем кофейни по районам и вычисляем статистику рейтингов
coffee_ratings_by_district = coffee_shops.groupby('district').agg(
    mean_rating=('rating', 'mean'),  # Средний рейтинг
    median_rating=('rating', 'median'),  # Медианный рейтинг
    count=('rating', 'count')  # Количество кофеен
).reset_index()

# Округляем
coffee_ratings_by_district['mean_rating'] = coffee_ratings_by_district['mean_rating'].round(2)
coffee_ratings_by_district['median_rating'] = coffee_ratings_by_district['median_rating'].round(2)

# Сортируем
coffee_ratings_by_district = coffee_ratings_by_district.sort_values(by='mean_rating', ascending=False)

coffee_ratings_by_district
```

```python
# График распределения рейтингов кофеен по районам
fig = px.box(
    coffee_shops,
    x='district',
    y='rating',
    color='district',
    title='Распределение рейтингов кофеен по районам',
    labels={'district': 'Район', 'rating': 'Рейтинг'},
    template='plotly_white',
    category_orders={'district': coffee_shops['district'].value_counts().index}  # Сортировка районов
)

fig.update_layout(
    xaxis_title='Район',
    yaxis_title='Рейтинг',
    xaxis_tickangle=45,
    showlegend=False
)

fig.show()
```

```python
# Отобразим рейтинг кофеен по округам на карте
# Создаём карту
m6 = Map(location=[moscow_lat, moscow_lng], zoom_start=10)

# Создаём хороплет
Choropleth(
    geo_data=state_geo,
    data=coffee_ratings_by_district,
    columns=['district', 'mean_rating'],
    key_on='feature.name',
    fill_color='YlGn',
    fill_opacity=0.8,
    line_opacity=0.2,
    legend_name='Средний рейтинг кофеен по округам'
).add_to(m6)

# Отображаем карту
m6
```

---

**Вывод:**<br/>
- Разница между средними рейтингами по округам минимальна (от 4.20 до 4.34), что указывает на относительно высокое и стабильное качество кофеен по всей Москве.
- **Лидер рейтинга:** Центральный административный округ (ЦАО) занимает первое место с самым высоким средним рейтингом кофеен — **4.34**.
- **Округа с высоким рейтингом:** СЗАО — **4.33**, САО — **4.29**.
- **Средние значения:** ВАО, ЮЗАО, ЮВАО, ЮАО имеют средний рейтинг около **4.22–4.28**.
- **Низший рейтинг:** ЗАО имеет самый низкий средний рейтинг — **4.20**.

### Стоимость чашки кофе

Определим на какую стоимость чашки капучино стоит ориентироваться при открытии и почему.

```python
# Рассчитаем медиану стоимости чашки капучино для всех кофеен
overall_median_price = coffee_shops['middle_coffee_cup'].median()

# Рассчитаем медиану стоимости чашки капучино по округам
median_price_by_district = (
    coffee_shops.groupby('district')['middle_coffee_cup']
    .median()
    .reset_index()
    .rename(columns={'middle_coffee_cup': 'median_coffee_price'})
    .sort_values(by='median_coffee_price', ascending=False)
)

print(f"Общая медиана стоимости чашки капучино: {overall_median_price:.2f} руб.")
print("Медиана стоимости чашки капучино по округам:")
print(median_price_by_district)
```

```python
# График медианной стоимости чашки капучино по округам
fig = px.bar(
    median_price_by_district,
    x='district',
    y='median_coffee_price',
    color='median_coffee_price',
    text='median_coffee_price',
    title='Медианная стоимость чашки капучино по округам',
    labels={'district_abbr': 'Округ', 'median_coffee_price': 'Медианная цена (руб.)'},
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_traces(textposition='outside')
fig.update_layout(title_x=0.5)

fig.show()
```

```python
# Отобразим медианную стоимость чашки кофе по округам на карте
# Создаём карту
m7 = Map(location=[moscow_lat, moscow_lng], zoom_start=10)

# Создаём хороплет
Choropleth(
    geo_data=state_geo,
    data=median_price_by_district,
    columns=['district', 'median_coffee_price'],
    key_on='feature.name',
    fill_color='YlOrRd',
    fill_opacity=0.8,
    line_opacity=0.2,
    legend_name=' по округам'
).add_to(m7)

# Отображаем карту
m7
```

---

**Вывод:**<br/>
- Самая дорогая чашка кофе продается в **ЮЗАО**, её средняя стоимость составляет **198 руб.**.
- Следующими по цене идут **ЦАО** и **ЗАО**, где чашка кофе стоит **190 руб.** и **189 руб.** соответственно.
- Самую низкую цену на кофе предлагает **ВАО**, где стоимость чашки составляет **135 руб.**.
- Ценовое распределение остаётся схожим с тем, что наблюдается при сравнении медианных цен по всем заведениям.
